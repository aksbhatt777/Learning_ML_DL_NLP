```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
%matplotlib inline 
```

```python
from sklearn.datasets import load_iris
iris=load_iris()
```

```python
iris
```

```python
print(iris['DESCR'])
```

```python
df=pd.DataFrame(iris['data'],columns=['sepal length in cm','sepal width','petal length','petal width'])
```

```python
df
```

```python
X=df ## independedt feature 
y=iris['target'] ## dependent feature 
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split as tts
X_train,X_test,y_train,y_test=tts(X,y,random_state=42,test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier as dtc
```

```python
dtc=dtc()
```

```python
dtc
```

```python
dtc.fit(X_train,y_train)
```

```python
## visualization of the tree 
from sklearn import tree
```

```python
plt.figure(figsize=(15,10))
tree.plot_tree(dtc,filled=True)
```

```python
y_pred=dtc.predict(X_test)
from sklearn.metrics import confusion_matrix,classification_report
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

## hyper parameter tuning -- prepruning 

```python
param={
    'criterion':['gini','entropy', 'log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['auto','sqrt','log2']
}
param
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
tc=DecisionTreeClassifier()
```

```python
grid=GridSearchCV(tc,param_grid=param,cv=5,scoring='accuracy')
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)
from sklearn.metrics import confusion_matrix,classification_report
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

### hyper parameters reduced the accurancu -- > the model was earlier overfitted
