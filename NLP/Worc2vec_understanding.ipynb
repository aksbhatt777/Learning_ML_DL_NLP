## SPAM HAM CLASSIFICATION 

```python
import pandas as pd
msg=pd.read_csv('/home/ubuntu/ds/nlp/SMSSpamCollection.txt',sep='\t',names=["label","message"])
```

```python
msg
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lem=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(0,len(msg)):
    x=re.sub('[^a-zA-z]',' ',msg['message'][i])
    x=x.lower()
    x=x.split()
    x=[lem.lemmatize(word) for word in x if not word in stopwords.words('english')]
    x=' '.join(x)
    corpus.append(x)
```

```python
[[i,j,k] for i,j,k in zip(list(map(len,corpus)),corpus, msg['message']) if i<1]
```

```python
[[i,j,k] for i,j,k in zip(list(map(len,corpus)),corpus, msg['message']) if i<1]
```

```python
## in word 2 vec --> vectorization happens for all the words .. at the moment corpus is list of sentences preprocessed
```

```python
from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess
from nltk.tokenize import word_tokenize
```

```python
words=[]
for sent in corpus:
    sent_token=sent_tokenize(sent)
    for word in sent_token:
        words.append(simple_preprocess(word))
```
