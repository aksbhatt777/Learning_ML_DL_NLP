```python
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

### This library does not create one-hot vectors. It converts words into integer IDs using a hashing trick.

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good scientist',
     'understand the meaning of words',
     'your videos are good',]
```

```python
type(sent)
```

```python
for words in sent:
    print(words)
    print(type(words))
```

### Converts a text into a list of integers, where each integer represents a word's index in a fixed-size vocabulary
### sequences = one_hot(text, n)
#### text: The input text to be encoded.
#### n: The size of the vocabulary.

```python
one_hot('word',10000) ## hashes the value of word within an interger of vocal till 10000 range 
```

```python
docs=[]
for i in sent:
    docs.append(one_hot(i, 10000))
```

```python
docs
```

### Word embedding 

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np 
```

```python
embed_doc=pad_sequences(docs,padding='post',maxlen=7)
embed_doc
```

```python
# This way we made all the sentences of equal length 
```

### Feature representation 

```python
model
```

```python
# Embedding turns nonnegative integers (indexes) into dense vectors of fixed size.
# Embedding(input_dim, output_dim)
model=Sequential()
model.add(Embedding(input_dim=10000,output_dim=10))
model.build((None, 7))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embed_doc)
```

```python
embed_doc[0]
```

```python
## we find for the first sentence embedding ... 7 words for each line ... each word is embedded by 10 values in the array 
model.predict(embed_doc)[0] #'the glass of milk'  --> [2043, 9757, 5924, 9341,    0,    0,    0]
#notice last 3 arrays shoudl be same as it is zero 
```

## WHAT HAPPENED HERE ?

1. All sentences hashed into integers using keras -- act as IDs
2. all sentences made into equilength -- called padding
3. embedding creates a dictionary for each word ID
   
   Word ID 1 -> [0.12, 0.83, ...]
Word ID 2 -> [0.91, 0.44, ...]
Word ID 3 -> [0.07, 0.66, ...]
...
Word ID 2043 -> [0.52, 0.19, ...]

4. Replace each ID with its vector... like a dictionary lookup
5. Then make predictions
