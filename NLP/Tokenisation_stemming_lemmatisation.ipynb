## STEMMING

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
from nltk.stem.porter import *
```

```python
stemmer = PorterStemmer()
```

```python
singles = [stemmer.stem(word) for word in words]
```

```python
singles
```

```python
for word in words:
    print(word+"  ------->  "+ stemmer.stem(word))
```

#### doing for individual words 

```python
stemmer.stem("gossiping"),stemmer.stem("singing"),stemmer.stem("operating")
```

#### Regexstemmer 

```python
from nltk.stem import RegexpStemmer
```

```python
regst = RegexpStemmer('ing$|s$|e$|able$|ion$', min=4)
```

```python
regst.stem('cars')
```

```python
regst.stem('communication')
```
