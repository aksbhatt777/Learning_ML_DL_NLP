### Snowball stemmer 

```python
para
```

```python
len(para), type(para)
```

```python
sentences=nltk.sent_tokenize(para)
```

```python
sentences
```

```python
len(sentences), type(sentences)
```

```python
from nltk.stem import SnowballStemmer
snow=SnowballStemmer('english')
```

```python
snow_naruto=[]
```

```python
for i in range(len(sentences)):
    words=word_tokenize(sentences[i])
    words=[snow.stem(word) for word in words if word not in stopwords.words('english')]
    snow_naruto.append(' '.join(words))
    
```

```python
snow_naruto
```

### lemmatize

```python
from nltk.stem import WordNetLemmatizer
wnl=WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words=word_tokenize(sentences[i])
    words=[wnl.lemmatize(word) for word in words if word not in stopwords.words('english')]
    sentences[i]=' '.join(words)
    
```

```python
sentences
```
